```python

import numpy as np

# Fix compatibility with older geoplotlib
np.int = int
np.float = float

import geoplotlib
import pandas as pd
from geoplotlib.utils import read_csv

print("Import successful!")
```

```python
dataset = read_csv('poaching_points_cleaned.csv')
dataset
```

```python
# Load the dataset using pandas
pd_dataset = pd.read_csv('poaching_points_cleaned.csv')
# Display the first five rows
pd_dataset.head()
```

```python
geoplotlib.dot(dataset)
geoplotlib.show()
```

```python
# Plotting poaching incidents as a histogram
geoplotlib.hist(dataset, binsize=20)
geoplotlib.show()
```

```python
# Plotting a Voronoi map
geoplotlib.voronoi(
    dataset,
    cmap='Blues_r',
    max_area=1e5,
    alpha=255
)

geoplotlib.show()
```
