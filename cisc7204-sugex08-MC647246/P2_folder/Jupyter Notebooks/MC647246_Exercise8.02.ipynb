```python
import numpy as np
if not hasattr(np, 'int'):
    np.int = int
if not hasattr(np, 'float'):
    np.float = float
import geoplotlib
from pathlib import Path

def data_file(filename):
    candidates = [Path('../Datasets') / filename, Path('Datasets') / filename, Path(filename)]
    for path in candidates:
        if path.is_file():
            return str(path)
    raise FileNotFoundError(f"Place {filename} in ../Datasets, Datasets, or this notebook folder")
import json
from geoplotlib.colors import ColorMap
from geoplotlib.utils import BoundingBox
```

```python
with open(data_file('us_states.json'), encoding='utf-8') as data:
    dataset = json.load(data)

fourth_state = json.loads(json.dumps(dataset['features'][3]))
coords = fourth_state['geometry']['coordinates']
while isinstance(coords, list) and coords and isinstance(coords[0], list):
    coords = coords[0]
fourth_state['geometry']['coordinates'] = coords
print(json.dumps(fourth_state, indent=4))
```

```python
with open(data_file('us_states.json'), encoding='utf-8') as data:
    dataset = json.load(data)

states = [feature['properties']['NAME'] for feature in dataset['features']]
print('Number of states:', len(states))
print(states)
```

```python
geoplotlib.geojson(data_file('us_states.json'))
geoplotlib.show()
```

```python
cmap = ColorMap('Reds', alpha=255, levels=10)

def get_color(properties):
    return cmap.to_color(properties['CENSUSAREA'], maxvalue=300000, scale='lin')
```

```python
geoplotlib.geojson(data_file('us_states.json'), fill=True, color=get_color)
geoplotlib.geojson(data_file('us_states.json'), fill=False, color=[0, 0, 0, 255])
geoplotlib.set_bbox(BoundingBox.USA)
geoplotlib.show()
```

```python
geoplotlib.geojson(
    data_file('us_states.json'), fill=True, color=get_color,
    f_tooltip=lambda properties: properties['NAME'] + ' - Census Areas: ' + str(properties['CENSUSAREA'])
)
geoplotlib.geojson(data_file('us_states.json'), fill=False, color=[0, 0, 0, 255])
geoplotlib.set_bbox(BoundingBox.USA)
geoplotlib.show()
```
